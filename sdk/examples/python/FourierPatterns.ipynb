```python
queue = laser_lib.DacQueue()
```

```python
# Circle
queue.dac_rate = 30000
T = 500
arr_pos = np.zeros((T, 2))
arr_col = np.zeros((T, 3))
arr_col[:, :] = 1
theta = np.linspace(0, 2*np.pi, T)
arr_pos[:, 0] = 0.5*np.cos(theta)
arr_pos[:, 1] = 0.5*np.sin(theta)

#arr_col = arr_col*np.expand_dims((theta > np.pi/2), 1)

while(True):
    queue.submit(arr_pos, arr_col, loop=False)
queue.submit(arr_pos, np.zeros((T,3)), loop=True)
```

```python
# Circle
queue.dac_rate = 50000
T = 1000
arr_x = np.linspace(-0.5, 0.5, T//2)
arr_col = np.zeros((T//2, 3))
arr_col_inv = np.zeros((T//2, 3))
arr_pos = np.zeros((T//2, 2))
arr_pos_inv = np.zeros((T//2, 2))

arr_pos[:, 0] = arr_x
arr_pos_inv[:, 0] = arr_x[::-1]


num_dots_inarow = 10
lin = np.linspace(1, T, T//2)
arr_col[:, :] = ((lin%num_dots_inarow*2) <= num_dots_inarow)[..., None]
arr_col[:, :] = arr_col[::-1, :]

while(True):
    for idx, y in enumerate(np.linspace(-0.5, 0.5, T//4)):
        if(idx % 2 == 0):
            arr_pos[:, 1] = y
            queue.submit(arr_pos, arr_col, loop=False)
        else:
            arr_pos_inv[:, 1] = y
            queue.submit(arr_pos_inv, arr_col, loop=False)
        
queue.submit(arr_pos, np.zeros((T,3)), loop=True)
```
